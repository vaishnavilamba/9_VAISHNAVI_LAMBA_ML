Basic tools you need.

requests → talks to the API
pandas → creates DataFrame

```python
import pandas as pd
import requests
```

```python
url = 'https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page=1'
```

Call the API

requests.get() sends a GET request
API replies with status code + data

```python
response = requests.get(url)
```

Check response status (VERY IMPORTANT)

200 → success ✅
404 → wrong URL ❌
500 → server issue ❌

```python
print(response.status_code)
```

Convert response to JSON

response.json() returns: list of dictionaries OR
dictionary

```python
data = response.json()
data
```

Convert JSON → DataFrame

```python
df = pd.DataFrame(data)
df.head()
```

the size of the DataFrame.

```python
df.shape
```

It saves DataFrame to a CSV file named movie.csv.

```python
df.to_csv('movie.csv')
```
