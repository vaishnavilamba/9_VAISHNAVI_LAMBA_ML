Bivariate Analysis means studying the relationship between two variables.

Example in house dataset:

Area vs Price → Does price increase when area increases?

Bedrooms vs Price → Do houses with more bedrooms cost more?

Stories vs Price → Does number of floors affect price?

```python
import pandas as pd
import seaborn as sns

print("Working")
```

Now dataset stored inside house variable.

```python

house = pd.read_csv('Housing.csv')
```

```python
house.head()
```

sns.scatterplot() → Draw dots graph

x = area

y = price.

```python
sns.scatterplot(x=house['area'],y=house['price'])
```

Shows average price per bedroom count
Shows relationship between:

Bedrooms ↔ Price

```python
sns.barplot(x=house['bedrooms'],y=house['price'])
```

Shows:

Minimum price

Maximum price

Average price

For each story count.

```python
sns.boxplot(x=house['stories'],y=house['price'])
```


house.corr() → Finds relationship between columns

Example output:

	Price	Area
Price	1	0.85
Area	0.85	1
Interpretation

0.85 = Strong relation

1 = Perfect relation

0 = No relation


```python
sns.heatmap(pd.crosstab(house['area'],house['price']))
```

Shows:

Area vs Price

Bedrooms vs Price

Bathrooms vs Price

All graphs together.

```python
sns.pairplot(house)
```

A lineplot shows the relationship between two numerical variables using a line.

Example:

Area vs Price

Area vs Bedrooms

Price vs Bathrooms

It connects points with a line.

Used when:

Data has an order

You want to see trend

```python
sns.lineplot(x=house['area'],y=house['price'])
```

```python
house_sorted = house.sort_values('area')
```

```python
sns.lineplot(x=house_sorted['area'],y=house_sorted['price'])
```
